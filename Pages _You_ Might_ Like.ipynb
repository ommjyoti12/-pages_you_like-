```python
import json
def data_covert(filename):
    with open(filename,"r") as f:
        return json.load(f) 
def pages_you_like(user_id,data):
    page_suggetion={}
    for user in data["users"]:
        page_suggetion[user['id']]=set(user["liked_pages"])

        if user_id not in  page_suggetion:
            return[]
        
    like_pages=   page_suggetion[user_id]
    page_suggestions={}
    for other_id,likes in  page_suggetion.items():
        if other_id != user_id:
            same_like=like_pages.intersection(likes)
            for page in likes:
                 if page not in  like_pages:
                    page_suggestions[page] = page_suggestions.get(page, 0) + len( same_like)
    
    # Sort recommended pages based on the number of shared interactions
    sorted_pages = sorted(page_suggestions.items(), key=lambda x: x[1], reverse=True)
    return [page_id for page_id, _ in sorted_pages]


data=data_covert("data2.json")
user_id=1
er=pages_you_like(user_id,data)
print(er)

               
                
    
        
      
        
        
        
   
        
        
    
```
